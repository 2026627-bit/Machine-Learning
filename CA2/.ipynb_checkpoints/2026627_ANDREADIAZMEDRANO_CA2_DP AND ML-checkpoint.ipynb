```python
missing_values_types=["n.a.", "?", "n/a","NA", "Na","UnKnown", "--", "na", "NONE", "Unknown", "-", "unknown", "none"]
df= pd.read_excel("CA2_Dataset.xlsx", na_values=missing_values_types)
df_clean=df.copy()
```

### TARGET VARIABLES


```python
df["Solar_PV_Demand_Level"].unique()
```

##### According to the data dictionary
Expected values for Solar_PV_Demand_Level:
    -Low
    -Average
    -High
As we have seen, there are different classifications, so we need to assign the correct category, adhering strictly to the definitions provided in the dictionary.


We will subsequently confirm the change that has been made.

```python
SP_replace={'Very High':'High', 'high':'High', 
          'Medium':'Average'}

df_clean['Solar_PV_Demand_Level']=df_clean['Solar_PV_Demand_Level'].replace(SP_replace)
```

```python
df_clean["Solar_PV_Demand_Level"].unique()
```

```python
df_clean["Solar_PV_Demand_Level"].isnull().sum()
```

```python
df["Projected_Solar_PV_Installation_Capacity_kWp"].unique()
```

```python
df["Projected_Solar_PV_Installation_Capacity_kWp"].describe()
```

##### According to the data dictionary
Expected values for Projected_Solar_PV_Installation_Capacity_kWp:
    Positive values; typically about 2 - 500 kWp, with substantial variation by customer segment

As we saw, the minimum value obtained was -25.00, which indicates that we have values outside the expected range; the same applies to the maximum value, which is 2500, even though the maximum should be 500.

```python
out_of_range=(
    (df_clean["Projected_Solar_PV_Installation_Capacity_kWp"]<0)|(df_clean["Projected_Solar_PV_Installation_Capacity_kWp"]>500))
print(out_of_range.sum())
```

##### Looking at the dictionary, we see that we have a value of -25, a 0, and values of 900 and 2500
Given the above, we will replace the values -25, 900, and 2500 with NaN values, and subsequently remove them.

```python
df_clean.loc[out_of_range, "Projected_Solar_PV_Installation_Capacity_kWp"]=np.nan
```

```python
df_clean["Projected_Solar_PV_Installation_Capacity_kWp"].isnull().sum()
```

### CATEGORICAL VARIABLES

```python
df["County"].unique()
```

##### According to the data dictionary
Expected values for Conunty:
    26 counties of the Republic of Ireland

As we saw, some counties are written in different ways, such as "Co Cork", while others contain spelling errors, like "Wexfrd" all of which need to be corrected to their proper names.

Subsequently, any empty values will be replaced with "Unknown."

```python
c_replace={'Dublín':'Dublin', 'Dubln':'Dublin', 
          'Co Cork':'Cork', 'Wexfrd':'Wexford'}

df_clean['County']=df_clean['County'].replace(c_replace)
```

```python
df_clean["County"].unique()
```

```python
df_clean["County"]=df_clean["County"].fillna("Unknown")
```

```python
df_clean["County"].isnull().sum()
```

### NUMERICAL VARIABLES

```python
df["Forecast_Year"].unique()
```

##### According to the data dictionary
Expected values for Forecast_Year:
    values between: 2027 - 2034

As we saw, this variable has wrong values for example 9999, 2050, 2045 or 2022. Aditionally has missing values.

For this reason, we will set these out-of-range values to NaN and subsequently impute them using the mode. We chose the mode rather than the mean because we are dealing with integers; furthermore, the number of missing values for this variable is 25, representing 1.2% of the total dataset (2,080 observations), so the risk of introducing bias into the distribution of the years is low.

```python
invalid_years= ((df_clean['Forecast_Year']<2027)| (df_clean['Forecast_Year']>2034))
df_clean.loc[invalid_years, 'Forecast_Year']=np.nan
```

```python
df_clean["Forecast_Year"]= df_clean["Forecast_Year"].fillna(round(df_clean['Forecast_Year'].median())).astype(int)
```

```python
df_clean["Forecast_Year"].unique()
```

```python
df_clean["Forecast_Year"].isnull().sum()
```

```python
df["Local_Mean_Household_Disposable_Income_EUR"].unique()
```

```python
df["Local_Mean_Household_Disposable_Income_EUR"].describe()
```

##### According to the data dictionary
Expected values for Local_Mean_Household_Disposable_Income_EUR:
    Typically EUR 35,000 - 125,000

As we can observe, the reference range in the dictionary spans from 35,000 to 125,000 euros. The detection of a minimum value of -12,000 suggests a recording error, as a negative average disposable income is economically impossible. Similarly, at the upper end—while the 125,000-euro limit is merely indicative ("typically") rather than a strict exclusion threshold—a maximum value of 250,000 (which is double that threshold and significantly distant from the 75th percentile of 87,290 euros) clearly constitutes an outlier. Consequently, both out-of-range values will be converted to NaN and subsequently imputed using the median, in order to preserve the integrity of the distribution.

```python
wrong_income= ((df_clean['Local_Mean_Household_Disposable_Income_EUR']<35000)| (df_clean['Local_Mean_Household_Disposable_Income_EUR']>125000))
df_clean.loc[wrong_income, "Local_Mean_Household_Disposable_Income_EUR"]=np.nan
```

```python
df_clean["Local_Mean_Household_Disposable_Income_EUR"]= df_clean["Local_Mean_Household_Disposable_Income_EUR"].fillna(df_clean['Local_Mean_Household_Disposable_Income_EUR'].median())
```
